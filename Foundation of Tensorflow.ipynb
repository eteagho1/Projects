What is tensor?
In Tensorflow all data is passed between operations in a computational graph and these are passed in form of tensors, hence the name tensorflow.
The word Tensors from the latin means "that which stretches".

```python
matrix_one = tf.constant([[1,3,4],[4,5,6],[8,7,9]])
matrix_two = tf.constant([[1,3,5],[4,7,6],[8,4,9]])
def matrix_add (matrix_one,matrix_two):
    add_operation = tf.add(matrix_one,matrix_two)
    mult_operation= tf.matmul(matrix_one,matrix_two)
    return add_operation,mult_operation
result=matrix_add(matrix_one,matrix_two)
tf.print(result)
```
